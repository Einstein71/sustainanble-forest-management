```python
import pandas as pd

df=pd.read_csv("baseball_positions.csv")
df["Position"]=df["Position"].str.split("/")
df_exploded= df.explode("Position")
df_exploded.groupby("Position")["Position"].count().sort_values(ascending=False)
```

```python
df = pd.read_csv("SPY_close_price_5Y.csv")
df['50_SMA'] = df['Close'].rolling(window=50, min_periods=1).mean()
df['200_SMA'] = df['Close'].rolling(window=200, min_periods=1).mean()
df.sort_values(by="Date", ascending=False)
df["golden_cross"] = df["50_SMA"] > df["200_SMA"]
df.head(50)
```

```python
df = pd.read_csv("LessonStreaks.csv")
df.sort_values("user_name", inplace=True)
# Keep only the first unique combination of user_name and date
df_unique = df.drop_duplicates(subset=["user_name", "date"])
df_unique.groupby("user_name")[["date", "lesson_id"]]
df
```
