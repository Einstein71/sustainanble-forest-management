```python
import kagglehub
path = kagglehub.dataset_download("shivamb/netflix-shows")
```

```python
import pandas as pd
import os
path_file=os.path.join(path, "netflix_titles.csv")
net_df=pd.read_csv(path_file)
net_df
net_df.groupby(["release_year", "type"])["show_id"].count().sort_values(ascending=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Filter the DataFrame to include only 'Movie' entries
movies_df = net_df[net_df['type'] == 'Movie']

# Group by release_year and count the number of movies for each year
movies_count_per_year = movies_df.groupby('release_year')['show_id'].count()

# Extract years and counts for plotting
x_ax = movies_count_per_year.index
y_ay = movies_count_per_year.values

# Create the plot
 # Set figure size for better readability
plt.plot(x_ax, y_ay, color="blue")
plt.xlabel("Release Year") # More descriptive label
plt.ylabel("Number of Movies") # Aligns with the filtered data
plt.title("Number of Movies Produced per Year")
plt.grid(True) # Add a grid for easier reading of values
plt.show()
```

```python
show_df=net_df[net_df["type"]=="TV Show"]
analyse_df=show_df.groupby("release_year")["show_id"].count()
a_ax=analyse_df.index
y_ay=analyse_df.values
plt.figure(figsize=(12,5))
plt.plot(a_ax, y_ay, color="blue")
plt.grid("show")
plt.xlabel("release year")
plt.ylabel("Number of tv shows")
plt.title("Tv series released yearly")
plt.show()
```

```python
data=analyse_df.values
labels=analyse_df.index
plt.pie(data, labels=labels, autopct="%1.1f%%", startangle=140)
centre_circle=plt.Circle((0,0), 0.80, fc="white")
fig=plt.gcf()
fig.gca().add_artist(centre_circle)
plt.legend(loc='best')
plt.axis("equal")
plt.show()
```
