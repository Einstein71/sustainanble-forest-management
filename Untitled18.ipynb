```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

```

```python
df=pd.read_excel('reltex.xlsx')
df.head()
```

```python
grouped_data=df.groupby('Month')[["Revenue_KES"]].sum().head()
grouped_data.head()
```

```python
labels= grouped_data.index
data=grouped_data.Revenue_KES.sort_values()
plt.pie(data, labels=labels, autopct="%1.1f%%", startangle=140)
plt.legend(loc="best")
plt.axis("equal")
plt.show()
```

```python
labels= grouped_data.index
data=grouped_data.Revenue_KES.sort_values()
plt.pie(data, labels=labels, autopct="%1.1f%%", startangle=140)
centre_circle= plt.Circle((0,0), 0.80, fc="white")
fig=plt.gcf()
fig.gca().add_artist(centre_circle)
plt.legend(loc="best")
plt.axis("equal")
plt.show()
```
