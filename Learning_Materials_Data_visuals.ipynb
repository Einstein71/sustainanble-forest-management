```python
import gradio as gr

# Define your core app logic
def simple_gsm_calculator(d,t):
  g=d/t/4.5
  # Round the result to two decimal places for the output
  round_g = round(g, 2)
  return round_g

# Create the UI layout with two numerical inputs
demo = gr.Interface(fn=simple_gsm_calculator,
                    inputs= [gr.Number(label="Denier"),
                             gr.Number(label="tape width")],
                    outputs=gr.Number(label="Gsm")) # Removed .2f from label and added rounding in the function

# Launch the app (share=True creates a public temporary URL)
demo.launch(share=True)
```

```python
def deforestation_report(**impact_value):
  print(f"deforestation_report:")
  for aspect in impact_value.items():
    value=impact_value[aspect]
    print(f'{aspect}:{value}')

    deforestation_report(
      loss_of_bio=30,
      carbon_emmission=500,
      soil_erosion=20)

    print(deforestation_report)
```

```python
def countdown(i):
  if i==0:
    print("Done")
  else:
    print(i)
    countdown(i-3)
countdown(30)
```

```python
def countd(i):
  if i==0:
    print("Done")
  else:
    print(i)
    countd(i-1)

countd(6)
```

```python
def factorial(n):
  if n==1:
    return n
  else:
    return n*factorial(n-1)

factorial(4)
```

```python
def fact(n):
  result=1
  count=2
  while count<=n:
    result=result*count
    count=count+1
  return result
fact(4)
```
